### Part 1: Reading and Displaying the Data
In this step, we open the file `my_data.txt` in read mode and print its contents to the screen.

````python
# Part 1: Ensure file exists and read content
import os

goals_text = """1. Learn to use the Python programming language confidently.
2. Learn how to use Python to solve mathematical problems.
3. Improve my understanding of basic Python concepts through practice.
4. Build a strong foundation for the Smart Finance Assistant project.
5. Successfully complete all lab assignments on time.
6. Manish Khatri - Semester Goals 2024"""

# Create file if it doesn't exist
if not os.path.exists('my_data.txt'):
    with open('my_data.txt', 'w') as f:
        f.write(goals_text)

# Read the content
with open('my_data.txt', 'r') as file:
    content = file.read()

print("--- Content of my_data.txt ---")
print(content)
print("------------------------------")
````

### Part 2: File Analysis
As required by the assignment, here are two pieces of information about the file: the total line count and the total word count.

````python
# Part 2: File Analysis
# Calculate the number of lines
lines = content.splitlines()
line_count = len(lines)

# Calculate the number of words
words = content.split()
word_count = len(words)

# Display the results for the assignment
print(f"--- Analysis for Manish Khatri ---")
print(f"Analysis Result 1: Number of lines = {line_count}")
print(f"Analysis Result 2: Number of words = {word_count}")
````

### Part 3: Data Visualization
To better understand my goals, I will create a simple bar chart that shows the word count for each goal. This helps identify which goals are more complex.

````python
import matplotlib.pyplot as plt

# Prepare the data for the chart using the 'lines' variable from the analysis step
goal_labels = [f"Goal {i+1}" for i in range(len(lines))]
word_counts = [len(line.split()) for line in lines]

# Create the bar chart
plt.figure(figsize=(10, 6))
plt.bar(goal_labels, word_counts, color='skyblue')

# Add titles and labels
plt.title('Word Count per Semester Goal', fontsize=14)
plt.xlabel('Goals', fontsize=12)
plt.ylabel('Number of Words', fontsize=12)

# Show the chart
plt.show()
````

## How to Upload Files to Google Colab

There are several ways to upload files to your Colab environment:

### Method 1: Using the File Browser (Recommended for small files)
1.  Click the **folder icon** on the left-hand sidebar to open the file browser.
2.  Click the **upload icon** (an arrow pointing upwards) in the file browser pane.
3.  Select the file(s) from your local computer that you wish to upload.
4.  The uploaded files will appear in the `/content/` directory.

### Method 2: Mounting Google Drive (Recommended for larger datasets or persistent storage)
1.  Run the following Python code in a cell:
    ```python
    from google.colab import drive
    drive.mount('/content/drive')
    ```
2.  A link will appear asking you to authorize Colab to access your Google Drive. Follow the instructions to grant permission.
3.  Once mounted, your Google Drive files will be accessible under `/content/drive/My Drive/`.

### Method 3: Uploading Programmatically (Using `files.upload()`)
1.  Run the following Python code in a cell:
    ```python
    from google.colab import files
    uploaded = files.upload()

    for fn in uploaded.keys():
      print(f'User uploaded file "{fn}" with length {len(uploaded[fn])} bytes')
    ```
2.  A button will appear allowing you to browse and select files from your local computer.
3.  This method is useful for interactive uploads and when you want to handle the uploaded file data directly in Python.
