# Week 6 – Worksheet: Mini Contact Book

## Part A — Basics (Must Do)

### A1. Create a dictionary with 3 contacts

````python
# Create a dictionary with 3 contacts
# Each name is the key and each phone number is a string

contacts = {
    "Alice": "0400 123 456",
    "Bob": "0411 000 999",
    "Charlie": "0422 555 888"
}

contacts
````

**What this does:**
Creates a dictionary called `contacts` with three people. The person's name is the key and their phone number is the value.

**Example output:**

```text
{'Alice': '0400 123 456',
 'Bob': '0411 000 999',
 'Charlie': '0422 555 888'}
```

---

### A2. Add a new entry

````python
# Add a new contact to the dictionary

contacts["David"] = "0433 111 222"

contacts
````

**What this does:**
Adds a new contact called David to the existing dictionary.

**Example output:**

```text
{'Alice': '0400 123 456',
 'Bob': '0411 000 999',
 'Charlie': '0422 555 888',
 'David': '0433 111 222'}
```

---

### A3. Update an existing contact’s number

````python
# Update Bob's phone number

contacts["Bob"] = "0455 777 333"

contacts
````

**What this does:**
Changes Bob's old phone number to a new phone number.

**Example output:**

```text
{'Alice': '0400 123 456',
 'Bob': '0455 777 333',
 'Charlie': '0422 555 888',
 'David': '0433 111 222'}
```

---

### A4. Print all contacts in a neat format

````python
# Loop through the dictionary and print each contact

for name, phone in contacts.items():
    print(name + ": " + phone)
````

**Example output:**

```text
Alice: 0400 123 456
Bob: 0455 777 333
Charlie: 0422 555 888
David: 0433 111 222
```

This completes the **required Part A checkpoint**:

* Dictionary created with 3+ contacts
* New contact added
* Existing contact updated
* All contacts printed neatly

# Part B — Nested (Stretch)

### B1. Convert contacts to a nested structure

````python
# Store both phone number and email for each contact

contacts = {
    "Alice": {
        "phone": "0400 123 456",
        "email": "alice@example.com"
    },
    "Bob": {
        "phone": "0455 777 333",
        "email": "bob@example.com"
    },
    "Charlie": {
        "phone": "0422 555 888",
        "email": "charlie@example.com"
    },
    "David": {
        "phone": "0433 111 222",
        "email": "david@example.com"
    }
}

contacts
````

**What this does:**
Each person's name is still the main key, but now each person has another dictionary containing their `phone` and `email`.

---

### B2. Print contacts neatly (name, phone, email)

````python
# Loop through the contacts and print their details

for name, details in contacts.items():
    print("Name:", name)
    print("Phone:", details["phone"])
    print("Email:", details["email"])
    print()
````

**Example output:**

```text
Name: Alice
Phone: 0400 123 456
Email: alice@example.com

Name: Bob
Phone: 0455 777 333
Email: bob@example.com

Name: Charlie
Phone: 0422 555 888
Email: charlie@example.com

Name: David
Phone: 0433 111 222
Email: david@example.com
```

---

### Optional lookup

````python
# Ask the user for a name and display their details

name = input("Enter a name: ")

if name in contacts:
    print("Name:", name)
    print("Phone:", contacts[name]["phone"])
    print("Email:", contacts[name]["email"])
else:
    print("Contact not found.")
````

**Example:**

```text
Enter a name: Alice
Name: Alice
Phone: 0400 123 456
Email: alice@example.com
```

The `if name in contacts` checks whether the name exists before trying to access the person's details.
